```python
# --- Colab: آپلود تصویر و جابه‌جایی کانال‌های رنگی ---
# این سلول را در گوگل کولب اجرا کنید.

from google.colab import files
import io, os
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
from pathlib import Path
from zipfile import ZipFile

# 1) آپلود تصویر
print("لطفاً یک تصویر آپلود کنید…")
uploaded = files.upload()
if not uploaded:
    raise SystemExit("هیچ فایلی آپلود نشد.")
in_name = next(iter(uploaded.keys()))
img = Image.open(io.BytesIO(uploaded[in_name])).convert("RGB")
arr = np.array(img)  # شکل: (H, W, 3) با ترتیب RGB

# ابزار کمکی برای جابه‌جایی دو کانال (با ایندکس‌های 0:R, 1:G, 2:B)
def swap_channels(rgb_array, i, j):
    out = rgb_array.copy()
    out[..., [i, j]] = out[..., [j, i]]
    return out

# 2) سه جابه‌جایی مستقل: R↔G ، R↔B ، G↔B
rg = swap_channels(arr, 0, 1)  # R<->G
rb = swap_channels(arr, 0, 2)  # R<->B
gb = swap_channels(arr, 1, 2)  # G<->B

# 3) جابه‌جایی‌های ترتیبی (یکی پس از دیگری): (R<->G) سپس (R<->B) سپس (G<->B)
seq = swap_channels(arr, 0, 1)  # R<->G
seq = swap_channels(seq, 0, 2)  # R<->B (روی تصویرِ قبلی)
seq = swap_channels(seq, 1, 2)  # G<->B (روی تصویرِ قبلی)

# 4) نمایش نتایج
titles = [
    "Original",
    "Swap R↔G",
    "Swap R↔B",
    "Swap G↔B"
]
images = [arr, rg, rb, gb, seq]

plt.figure(figsize=(15, 8))
for i, (title, im) in enumerate(zip(titles, images), start=1):
    plt.subplot(2, 3, i)
    plt.imshow(im)
    plt.axis("off")
    plt.title(title)
plt.tight_layout()
plt.show()

# 5) ذخیرهٔ فایل‌ها
base = Path(in_name).stem
os.makedirs("outputs", exist_ok=True)

paths = []
def save_img(np_img, suffix):
    p = Path("outputs")/f"{base}_{suffix}.png"
    Image.fromarray(np_img).save(p)
    paths.append(p)

save_img(arr, "original")
save_img(rg,  "swap_RG")
save_img(rb,  "swap_RB")
save_img(gb,  "swap_GB")

# یک فایل ZIP برای دانلود راحت‌تر
zip_path = Path("outputs")/f"{base}_channel_swaps.zip"
with ZipFile(zip_path, "w") as zf:
    for p in paths:
        zf.write(p, arcname=p.name)

print("فایل‌ها ذخیره شدند در مسیر outputs/. برای دانلود ZIP از سلول زیر استفاده کنید.")
```
